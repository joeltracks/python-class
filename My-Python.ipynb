### conditional statements
if true perform a task otherwise do nothing

```python
attained_grade = 1
target_grade = 51

if attained_grade == 0:
    print("no grade contact lecturer for assistance")
#check if target grade is achieved
elif attained_grade >= target_grade:
    print("pass")

#check if grade is less than target grade

elif attained_grade < target_grade:
    print("fail")

#if anything else apart from the above conditions, print "invalid grade"

else:
    print("invalid grade contact lecturer for assistance")
```

# LOOPS

we have:
- for loops
- while loops

```python
student_grades = [10, 13, 65, 59, 21, 45, 90, 33, 0, 88]
```

```python
print(student_grades)
```

```python
# printing the grae of each student on the list
for grade in student_grades:
    print(grade)
    
```

**conditional statements in for loops**

```python
# print only grades that are greater than or equal to 50

for grade in student_grades:
    #check if the grade is greater than or equal to 50
    if grade >= 50:
        print(grade)
```

```python
# check how many students have scored 50 and above
graduating_students = 0

for grade in student_grades:
    #check if the grade is greater than or equal to 50
    if grade >= 50:
        graduating_students += 1

print(f"Number of students with a score of 50 and above: {graduating_students}")
```

```python
# check how many students have scored 50 and above
graduating_students = 0
students_not_graduating = 0

for grade in student_grades:
    #check if the grade is greater than or equal to 50
    if grade >= 50:
        graduating_students += 1
    else:
        students_not_graduating += 1

print(f"Number of students graduating: {graduating_students}")
print(f"Number of students not graduating: {students_not_graduating}")
print(f"Total number of students: {graduating_students + students_not_graduating}")

```

# while loops

- useful for continous tasks


```python
# Items in stock
stock = 10

# Number of purchases
purchases = 0

while purchases < stock:
    # When a customer buys increase the number of purchases by 1
    purchases += 1
    #print remaining stock after each purchase
    print(f"Remaining stock: {stock - purchases}")
```
