```python
import numpy as np

# Read the CSV file. The column names become names we can use in NumPy.
data = np.genfromtxt(
    "real_estate_dataset.csv",
    delimiter=",",
    names=True,
    dtype=None,
    encoding="utf-8"
)

print("Number of homes:", len(data))
print("First home:")
print(data[0])

# Get a few columns from the table.
prices = data["Price"]
sizes = data["Square_Feet"]
print("\nFirst five prices:", prices[:5])
print("First five sizes:", sizes[:5])
```

```python
price_per_sqft = prices / sizes

print("PRICE (USD)")
print("Mean:", round(np.mean(prices), 2))
print("Median:", round(np.median(prices), 2))
print("Minimum:", round(np.min(prices), 2))
print("Maximum:", round(np.max(prices), 2))

print("\nSIZE (square feet)")
print("Mean:", round(np.mean(sizes), 2))
print("Median:", round(np.median(sizes), 2))
print("Minimum:", round(np.min(sizes), 2))
print("Maximum:", round(np.max(sizes), 2))

print("\nPRICE PER SQUARE FOOT (USD)")
print("Mean:", round(np.mean(price_per_sqft), 2))
print("Median:", round(np.median(price_per_sqft), 2))
print("Minimum:", round(np.min(price_per_sqft), 2))
print("Maximum:", round(np.max(price_per_sqft), 2))

# The 75th percentile is the price cutoff for the top quarter.
price_cutoff = np.percentile(prices, 75)
print("\n75th percentile price:", round(price_cutoff, 2))
```

```python
top_mask = prices >= price_cutoff
top_homes = data[top_mask]

print("Homes in the top quarter:", len(top_homes))
print("IDs of the first ten homes in that group:", top_homes["ID"][:10])

# Compare every home's price with the average price.
average_price = np.mean(prices)
above_average = prices > average_price
below_average = prices < average_price

print("\nAverage price:", round(average_price, 2))
print("Homes above average:", np.sum(above_average))
print("Homes below average:", np.sum(below_average))

# Check one example property: the first row in the data.
first_price = prices[0]
if first_price > average_price:
    print("The first home's price is above average.")
else:
    print("The first home's price is below or equal to average.")
```

```python
price_order = np.argsort(prices)
price_order = price_order[::-1]
top_ten = data[price_order[:10]]

print("ID | Size | Bedrooms | Location score | Price")
for home in top_ten:
    print(
        int(home["ID"]), "|",
        round(home["Square_Feet"], 2), "|",
        int(home["Num_Bedrooms"]), "|",
        round(home["Location_Score"], 2), "|",
        round(home["Price"], 2)
    )

print("\nAverage size of these homes:", round(np.mean(top_ten["Square_Feet"]), 2))
print("Average bedrooms of these homes:", round(np.mean(top_ten["Num_Bedrooms"]), 2))
print("Average location score of these homes:", round(np.mean(top_ten["Location_Score"]), 2))
```

## 5. Business insight
The ten highest-priced homes average 267.4 square feet and 4.7 bedrooms, compared with 174.6 square feet and 3.0 bedrooms across all homes. Their average location score is 6.6, compared with 5.2 overall. In this dataset, larger homes and homes with more bedrooms are associated with higher prices; the top ten also have a higher average location score.
