## 3. Operations and Calculated Columns

Two calculated columns were created to support the analysis.

- **Shipping Duration** was calculated as the number of days between the Order Date and Ship Date.

- **Profit Margin** was calculated as Profit divided by Sales and multiplied by 100. 

- The analysis also calculated total sales, total profit, total quantity sold, average discount, total shipping cost, average shipping duration, and average profit margin.

- These metrics provide an overall view of the company's sales performance, profitability, discounting, and shipping efficiency.

```python
orders["Shipping Duration"] = (
    orders["Ship Date"] - orders["Order Date"]
).dt.days
orders[["Order Date", "Ship Date", "Shipping Duration"]].head()
```

```python
orders["Profit Margin"] = (
    orders["Profit"] / orders["Sales"]
) * 100 
orders["Profit Margin"] = orders["Profit Margin"].round(2)
orders[["Sales", "Product Name", "Profit", "Profit Margin"]].head()
```

```python
total_sales = orders["Sales"].sum().round(2)
print("Total Sales:", total_sales)
total_profit = orders["Profit"].sum().round(2)
print("Total Profit:", total_profit)
total_quantity = orders["Quantity"].sum()
print("Total Quantity Sold:", total_quantity)
average_discount = orders["Discount"].mean().round(2)
print("Average Discount:", average_discount)
total_shipping_cost = orders["Shipping Cost"].sum().round(2)
print("Total Shipping Cost:", total_shipping_cost)
```

## 4. Combining the Datasets

- The Orders and Returns datasets were combined using the `merge()`
function and the `Order ID` column as the common key.

- A left join was used because the Orders dataset contains the complete
set of order records, while the Returns dataset only contains orders
that were returned. This ensures that all orders remain in the combined
dataset.

- Missing values in the `Returned` column represent orders that were not
listed in the Returns dataset. These values were therefore replaced with
"No", while existing returned orders retained the value "Yes".

```python
returns.head()
```

```python
orders_returns = orders.merge(
    returns,
    on="Order ID",
    how="left"
)
orders_returns.head()
```

```python
orders_returns["Returned"] = orders_returns["Returned"].fillna("No")
orders_returns["Returned"].value_counts()
```

```python
returned_percentage = (
    orders_returns["Returned"].eq("Yes").mean() * 100
)

print(f"Returned order lines: {returned_percentage:.2f}%")
```

```python
unique_orders = orders_returns["Order ID"].nunique()

returned_orders = orders_returns.loc[
    orders_returns["Returned"] == "Yes",
    "Order ID"
].nunique()

returned_order_percentage = (
    returned_orders / unique_orders
) * 100

print("Total unique orders:", unique_orders)
print("Unique returned orders:", returned_orders)
print(f"Percentage of orders returned: {returned_order_percentage:.2f}%")
```

```python
orders_combined = orders_returns.merge(
    people,
    left_on="Region_x",
    right_on="Region",
    how="left"
)
orders_combined.shape
```

## 4. Combining the Datasets

The Orders, Returns, and People datasets were combined using common keys.

The Orders and Returns datasets were first merged using `Order ID` with a left join. This preserved all order records while adding return information where available. Missing values in the `Returned` column were replaced with `"No"`.

The People dataset was then matched to the combined data using the regional information. During this process, a regional mismatch was identified for Canadian records. The Orders dataset used `"Canada"`, while the People dataset divided Canada into `"Eastern Canada"` and `"Western Canada"`.

The Canadian provinces were therefore mapped to the corresponding regions. Ontario, Quebec, Nova Scotia, and Newfoundland were classified as Eastern Canada, while Alberta, British Columbia, Saskatchewan, and Manitoba were classified as Western Canada.

After resolving this mismatch, salesperson information was successfully assigned to the Canadian records.

The final combined dataset retained all 51,290 order rows without unintended row multiplication.

```python
canada_region_map = {
    "Ontario": "Eastern Canada",
    "Quebec": "Eastern Canada",
    "Nova Scotia": "Eastern Canada",
    "Newfoundland": "Eastern Canada",
    "Alberta": "Western Canada",
    "British Columbia": "Western Canada",
    "Saskatchewan": "Western Canada",
    "Manitoba": "Western Canada"
}
```

```python
orders_combined.loc[
    orders_combined["Region_x"] == "Canada",
    "Region_x"
] = orders_combined.loc[
    orders_combined["Region_x"] == "Canada",
    "State"
].map(canada_region_map)
```

```python
orders_combined["Region_x"].value_counts().loc[
    ["Eastern Canada", "Western Canada"]
]
```

```python
id_to_person = people.set_index("Region")["Person"]

orders_combined["Person"] = orders_combined["Region_x"].map(id_to_person)
```

### Market Performance

The market-level analysis shows substantial differences in sales and profitability.

Asia Pacific generated the highest total sales at approximately **$4.04 million**, with 48,597 units sold. Europe generated approximately **$3.29 million in sales**, but recorded the highest total profit at approximately **$449,552**.

LATAM generated approximately **$2.16 million in sales and $221,643 in profit**, while USCA generated approximately **$2.36 million in sales and $304,214 in profit**.

Africa had the lowest overall performance, with approximately **$783,773 in sales and $88,872 in profit**.

The difference between sales and profit across markets suggests that revenue alone does not fully explain profitability. Further analysis of discounts, product categories, and regional performance may help explain these differences.

```python
market_summary = orders_combined.groupby("Market")[["Sales", "Profit", "Quantity"]].sum()

market_summary.round(2)
```

### Regional Performance

The regional analysis shows significant differences in both sales and profitability.

Western Europe was the strongest region, generating approximately **$1.73 million in sales and $218,434 in profit** from 22,262 units sold.

Eastern Asia and Southern Asia also performed strongly, generating approximately **$167,102** and **$159,336** in profit respectively.

However, several regions generated negative total profit. Western Asia recorded approximately **$317,107 in sales but a loss of $53,922**, while Western Africa generated approximately **$173,879 in sales but a loss of $50,408**. Central Asia also recorded a smaller loss of approximately **$7,282**.

Southeastern Asia generated relatively high sales of approximately **$884,423**, but only $17,852 in profit, indicating low profitability despite substantial sales volume.

These results demonstrate that high sales do not necessarily translate into high profitability. The loss-making regions should be investigated further by examining discounts, product categories, and individual products.

```python
region_summary = orders_combined.groupby("Region_x")[["Sales", "Profit", "Quantity"]].sum()

region_summary.sort_values("Profit", ascending=False).round(2)
```

### Category Performance

- Technology was the most profitable product category, generating approximately **$4.74 million in sales and $663,779 in profit** from 35,176 units sold.

- Office Supplies generated approximately **$3.79 million in sales and $518,596 in profit**. It had the highest sales volume, with **108,244 units sold**, showing that it is the largest category by quantity.

- Furniture generated approximately **$4.11 million in sales but only $285,083 in profit** from 34,892 units. Despite having higher sales than Office Supplies, Furniture generated substantially less profit.

- These results show that sales volume and revenue do not necessarily correspond to profitability. Technology appears to be the strongest category from a profit perspective.

```python
category_summary = orders_combined.groupby("Category")[["Sales", "Profit", "Quantity"]].sum()

category_summary.sort_values("Profit", ascending=False).round(2)
```

```python
subcategory_summary = orders_combined.groupby("Sub-Category")[[
    "Sales", "Profit", "Quantity"
]].sum()

subcategory_summary.sort_values("Profit", ascending=False).round(2)
```
